```python
import scanpy as sc
import pandas as pd
import os

SAMPLE_DIR = "/Users/adiallo/Desktop/Thesis/Data_Documents/All_Data/Visium_samples/Visium_Data/samples"
sids = ["0", "10", "11", "38", "80", "97"]

all_regions = []
for sid in sids:
    adata = sc.read_h5ad(os.path.join(SAMPLE_DIR, f"{sid}.h5ad"))
    df = pd.DataFrame({
        'barcode': adata.obs_names,
        'region': adata.obs['region'].values,
        'refined_region': adata.obs['refined_region'].values,
        'sample_id': sid
    })
    all_regions.append(df)
    print(f"  {sid}: {len(df)} spots")

regions = pd.concat(all_regions)
out = os.path.join(SAMPLE_DIR, "decon_benchmark", "region_lookup.csv")
os.makedirs(os.path.dirname(out), exist_ok=True)
regions.to_csv(out, index=False)
print(f"Saved {len(regions)} rows to {out}")
```
